# Task
Crear un software en Python para calcular el Índice de Masa Corporal (IMC) de pacientes en un consultorio médico, clasificar su estado (bajo peso, normal, sobrepeso, obesidad) y ofrecer recomendaciones personalizadas basadas en la clasificación.

## Obtener datos del paciente

### Subtask:
Solicitar al usuario el peso (en kilogramos) y la altura (en metros) del paciente.


**Reasoning**:
Request the patient's weight and height from the user and store them as float variables.



```python
peso = float(input("Ingrese el peso del paciente en kilogramos: "))
altura = float(input("Ingrese la altura del paciente en metros: "))
```

## Calcular el imc

### Subtask:
Implementar la fórmula del IMC: peso / (altura * altura).


**Reasoning**:
Calculate the IMC using the given formula and store the result in a variable.



```python
imc = peso / (altura * altura)
```

## Clasificar el estado del paciente

### Subtask:
Basándose en el valor del IMC, determinar si el paciente tiene bajo peso, peso normal, sobrepeso u obesidad.


**Reasoning**:
Implement conditional statements to classify the IMC value and store the classification in `estado_imc`.



```python
if imc < 18.5:
  estado_imc = "Bajo peso"
elif 18.5 <= imc < 25:
  estado_imc = "Peso normal"
elif 25 <= imc < 30:
  estado_imc = "Sobrepeso"
else:
  estado_imc = "Obesidad"
```

## Generar recomendaciones

### Subtask:
Proporcionar recomendaciones personalizadas basadas en la clasificación del IMC del paciente.


**Reasoning**:
Based on the classification of the IMC, create a series of conditional statements to provide personalized recommendations.



```python
if estado_imc == "Bajo peso":
  recomendaciones = "Se recomienda una evaluación nutricional para aumentar de peso de forma saludable."
elif estado_imc == "Peso normal":
  recomendaciones = "Continúe con un estilo de vida saludable para mantener su peso actual."
elif estado_imc == "Sobrepeso":
  recomendaciones = "Considere consultar a un profesional de la salud o nutricionista para un plan de pérdida de peso."
else: # estado_imc == "Obesidad"
  recomendaciones = "Es altamente recomendable buscar asesoramiento médico y nutricional para un plan integral de manejo de peso."
```

## Mostrar resultados

### Subtask:
Presentar el IMC calculado, la clasificación del estado del paciente y las recomendaciones.


**Reasoning**:
Print the calculated IMC, the classification, and the recommendations to the user.



```python

peso = float(input("Ingrese el peso del paciente en kilogramos: "))
altura = float(input("Ingrese la altura del paciente en metros: "))


imc = peso / (altura * altura)


if imc < 18.5:
  estado_imc = "Bajo peso"
elif 18.5 <= imc < 25:
  estado_imc = "Peso normal"
elif 25 <= imc < 30:
  estado_imc = "Sobrepeso"
else:
  estado_imc = "Obesidad"


if estado_imc == "Bajo peso":
  recomendaciones = "Se recomienda una evaluación nutricional para aumentar de peso de forma saludable."
elif estado_imc == "Peso normal":
  recomendaciones = "Continúe con un estilo de vida saludable para mantener su peso actual."
elif estado_imc == "Sobrepeso":
  recomendaciones = "Considere consultar a un profesional de la salud o nutricionista para un plan de pérdida de peso."
else:
  recomendaciones = "Es altamente recomendable buscar asesoramiento médico y nutricional para un plan integral de manejo de peso."


print(f"\nEl Índice de Masa Corporal (IMC) del paciente es: {imc:.2f}")
print(f"Clasificación del estado del paciente: {estado_imc}")
print(f"Recomendaciones: {recomendaciones}")
```
